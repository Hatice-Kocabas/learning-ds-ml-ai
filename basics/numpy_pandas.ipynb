# Titanic Dataset Exploratory Data Analysis (EDA) with NumPy & Pandas

   This notebook demonstrates basic data loading, cleaning, and exploration using Pandas and NumPy.

```python
# importing libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
# readingpublic titanic dataset using pandas from url
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
```

```python
# first five row
df.head()
```

```python
# statistics about row
df.info()
df.describe()
```

```python
# missing values
df.isnull().sum()
```

```python
# filling missing values with median in age column
df['Age'].fillna(df['Age'].median(), inplace=True)
```

```python
# Dropping columns from a dataset
df.dropna(subset=['Embarked'], inplace=True)
```

```python
# grouping pclass according to survived
df.groupby('Pclass')['Survived'].mean()

```

```python
#filtering dataframe where age bigger than 50
older_passengers = df[df['Age'] > 50]
older_passengers.head()
```

```python
#example of numpy functions
ages = df['Age'].to_numpy()
print("Mean Age:", np.mean(ages))
print("Median Age:", np.median(ages))
print("Standard Deviation of Age:", np.std(ages))
```
